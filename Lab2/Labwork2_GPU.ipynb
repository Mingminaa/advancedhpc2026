```python
from numba import cuda
cuda.detect()
```

```python
device = cuda.select_device(0)
print(f"ID: {device.id}")
print(f"Device: {device.name}")
```

```python
cc = device.compute_capability
sm_count = device.MULTIPROCESSOR_COUNT

cc_cores_per_SM = {
    (2,0) : 32,
    (2,1) : 48,
    (3,0) : 192,
    (3,5) : 192,
    (3,7) : 192,
    (5,0) : 128,
    (5,2) : 128,
    (6,0) : 64,
    (6,1) : 128,
    (7,0) : 64,
    (7,5) : 64,
    (8,0) : 64,
    (8,6) : 128,
    (8,9) : 128,
    (9,0) : 128,
    (10,0) : 128,
    (12,0) : 128
    }

core_per_SM = cc_cores_per_SM.get(cc)
print(f"Multiprocessor count: {sm_count}")
print(f"Total cores: {sm_count*core_per_SM}")
```

```python
meminfo = cuda.current_context().get_memory_info()
print(f"Free Memory: {meminfo[0]} bytes, Total Memory: {meminfo[1]} bytes")
```
